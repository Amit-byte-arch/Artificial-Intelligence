```python
#next we will see how we will use UNI-GRAM, BI- GRAM, TRI-GRAM USING NLTK
from nltk.util import bigrams, trigrams, ngrams
```

```python
string = 'Generative ai & agentic ai is automate the industry & more focus on every domain'
string
```

```python
quotes_tokens = nltk.word_tokenize(string) # the statement has converted into tokens
quotes_tokens
```

```python
len(quotes_tokens)
```

```python
quotes_bigrams = list(nltk.bigrams(quotes_tokens))
quotes_bigrams
```

```python
quotes_trigrams = list(nltk.trigrams(quotes_tokens))
quotes_trigrams
```

```python
quotes_ngrams = list(nltk.ngrams(quotes_tokens))
quotes_ngrams
```

```python
quotes_ngrams = list(nltk.ngrams(quotes_tokens, 4))
quotes_ngrams
#it has given n-gram of length 4
```

```python
quotes_ngrams = list(nltk.ngrams(quotes_tokens, 7))
quotes_ngrams
```

```python
quotes_ngrams = list(nltk.ngrams(quotes_tokens, 10))
quotes_ngrams
```

```python
quotes_ngrams = list(nltk.ngrams(quotes_tokens, 6))
quotes_ngrams
```

```python
from nltk.stem import PorterStemmer
pst = PorterStemmer()
```

```python
pst.stem('affection')
```

```python
pst.stem('playing')
```

```python
pst.stem('maximum')
```

```python
word_to_stem = ['give', 'giving', 'given', 'gave']

for words in word_to_stem:
    print(words+ ' : ' +  pst.stem(words))
```

```python
word_to_stem = ['give', 'giving', 'given', 'gaved', 'thinking', 'loving', 'maximum', 'Hello']

for words in word_to_stem:
    print(words+ ' : ' +  pst.stem(words))
```

```python
from nltk.stem import LancasterStemmer # remove the last suffix of every word
lst = LancasterStemmer()

for words in word_to_stem:
    print(words+ ' : ' +  lst.stem(words))
```

```python
from nltk.stem import SnowballStemmer
sbst = SnowballStemmer('english')

for words in word_to_stem:
    print(words+ ' : ' +  sbst.stem(words))
```
